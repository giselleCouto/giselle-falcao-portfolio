> **Quais estados brasileiros mais cresceram em população entre 2020 e 2025 — e o que isso diz para quem planeja serviços e negócios?**

## Etapa 2 — Obtenção dos dados

Os dados abaixo são uma **amostra didática**: valores aproximados e arredondados, para fins de estudo, **inspirados nas estimativas populacionais do IBGE** (o instituto oficial de estatística do Brasil — ibge.gov.br).

**Pergunta de ouro da granularidade:** cada linha desta tabela representa o quê? *Resposta: uma unidade da federação (UF).* São 27 linhas — 26 estados + o Distrito Federal — com 5 colunas: sigla, região, população estimada em 2020 e em 2025 (em **milhares de habitantes**) e área em km².

```python
# Montamos a tabela (o pandas chama de DataFrame) a partir de listas.
# Não precisa entender cada detalhe do código: foque na estrutura dos dados.

dados = {
    "uf": ["SP", "MG", "RJ", "BA", "PR", "RS", "PE", "CE", "PA", "SC",
           "GO", "MA", "AM", "PB", "ES", "MT", "RN", "AL", "PI", "DF",
           "MS", "SE", "RO", "TO", "AC", "AP", "RR"],
    "regiao": ["Sudeste", "Sudeste", "Sudeste", "Nordeste", "Sul", "Sul",
               "Nordeste", "Nordeste", "Norte", "Sul", "Centro-Oeste",
               "Nordeste", "Norte", "Nordeste", "Sudeste", "Centro-Oeste",
               "Nordeste", "Nordeste", "Nordeste", "Centro-Oeste",
               "Centro-Oeste", "Nordeste", "Norte", "Norte", "Norte",
               "Norte", "Norte"],
    "pop_2020_mil": [44000, 20540, 16550, 14140, 11080, 10880, 9050, 8710,
                     8070, 7000, 6920, 6680, 3940, 3940, 3830, 3440, 3300,
                     3220, 3190, 2900, 2710, 2220, 1700, 1510, 830, 770, 530],
    "pop_2025_mil": [46000, 21400, 17100, 14900, 11900, 11300, 9500, 9200,
                     8700, 8100, 7400, 7000, 4300, 4150, 4100, 3900, 3450,
                     3230, 3290, 2980, 2900, 2290, 1750, 1580, 880, 800, 720],
    "area_km2": [248219, 586513, 43750, 564760, 199298, 281707, 98067,
                 148894, 1245870, 95730, 340125, 329642, 1559167, 56467,
                 46074, 903207, 52809, 27843, 251756, 5760, 357145, 21926,
                 237765, 277720, 164173, 142470, 224273]
}

df = pd.DataFrame(dados)

print(f"Tabela criada: {df.shape[0]} linhas (UFs) e {df.shape[1]} colunas.")
```

```python
# head() mostra as 5 primeiras linhas da tabela — sempre o primeiro olhar.
# Confira: as colunas fazem sentido? Os valores parecem razoáveis?

df.head()
```

## Etapa 3 — Limpeza e exploração

Nossa amostra didática já chegou limpa (na vida real, raramente chega — você viu a fama dos 80% na Aula 1.3). Então vamos direto à **exploração**: conhecer os dados antes de confiar neles.

A célula abaixo usa `describe()`, que gera um resumo estatístico de cada coluna numérica: contagem, média, mínimo, máximo e os quartis (você vai dominar cada um deles no Módulo 2).

**Enquanto executa, anote mentalmente:** qual é a maior e a menor população de 2025? A média das populações descreve bem um "estado típico"? *(Repare como São Paulo, sozinho, puxa a média para cima...)*

```python
# describe() = raio-X estatístico das colunas numéricas.
# round(1) apenas arredonda para facilitar a leitura.

df[["pop_2020_mil", "pop_2025_mil", "area_km2"]].describe().round(1)
```

```python
# Agora criamos duas colunas novas — o coração da investigação:
# crescimento_mil -> quantas MIL pessoas o estado ganhou (absoluto)
# crescimento_pct -> quanto isso representa do tamanho que ele tinha (percentual)

df["crescimento_mil"] = df["pop_2025_mil"] - df["pop_2020_mil"]
df["crescimento_pct"] = (df["crescimento_mil"] / df["pop_2020_mil"] * 100).round(1)

# sort_values ordena a tabela; ascending=False = do maior para o menor.
top10_pct = df.sort_values("crescimento_pct", ascending=False).head(10)

print("TOP 10 — crescimento PERCENTUAL (2020 a 2025):")
top10_pct[["uf", "regiao", "pop_2020_mil", "pop_2025_mil", "crescimento_mil", "crescimento_pct"]]
```

### Pare e interprete: absoluto × percentual

Execute mentalmente esta comparação (ou adicione `df.sort_values("crescimento_mil", ascending=False).head(10)` em uma célula nova para conferir):

- No ranking **absoluto**, quem lidera é **São Paulo**: cerca de 2 milhões de novos habitantes — mais gente que a população inteira de vários estados.
- No ranking **percentual**, quem lidera é **Roraima**: os mesmos ~190 mil novos habitantes que passariam despercebidos em SP representam um salto de mais de 30% para o estado.

**Nenhum ranking está "errado"** — eles respondem perguntas diferentes. O absoluto dimensiona o volume (quantas escolas e hospitais a mais?); o percentual mostra o ritmo (onde a estrutura atual vai ficar apertada mais rápido?). Uma boa análise apresenta os dois.

## Etapa 4 — Análise: e por região?

Estados são 27; regiões são 5 — agregar por região resume a história. A célula abaixo usa `groupby`, o "agrupe e calcule" do pandas (parente da tabela dinâmica das planilhas).

```python
# Para cada região: soma das populações e média do crescimento percentual.

por_regiao = df.groupby("regiao").agg(
    pop_2025_mil=("pop_2025_mil", "sum"),
    crescimento_pct_medio=("crescimento_pct", "mean")
).round(1).sort_values("crescimento_pct_medio", ascending=False)

por_regiao
```

```python
# Um gráfico vale por dez tabelas — desde que seja honesto e legível.
# Repare no título: ele AFIRMA o achado (técnica do Módulo 4).

top10_ordenado = top10_pct.sort_values("crescimento_pct")  # menor embaixo, maior no topo

plt.figure(figsize=(9, 5))
plt.barh(top10_ordenado["uf"], top10_ordenado["crescimento_pct"], color="#7C5CBF")
plt.title("Roraima cresce mais de 30% e lidera o ritmo populacional do país (2020–2025)")
plt.xlabel("Crescimento populacional (%)")
plt.ylabel("UF")
plt.tight_layout()
plt.show()
```

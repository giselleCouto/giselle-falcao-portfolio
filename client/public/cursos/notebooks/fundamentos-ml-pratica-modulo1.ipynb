```python
# Importando as bibliotecas que vamos usar.
# Elas já vêm instaladas no Google Colab — nenhum download é necessário.

import io                          # para ler o texto CSV embutido neste notebook
import pandas as pd                # pandas: a biblioteca para trabalhar com tabelas de dados
import matplotlib.pyplot as plt    # matplotlib: a biblioteca para criar gráficos

print("Bibliotecas importadas com sucesso! Versão do pandas:", pd.__version__)
```

## 1. Carregando os dados

Nosso dataset tem **uma linha por fazenda** e as seguintes colunas:

| Coluna | O que é |
|--------|---------|
| `fazenda` | código da fazenda (F01 a F24) |
| `municipio` | município da fazenda em Minas Gerais |
| `n_vacas` | número de vacas em lactação |
| `racao_kg_vaca_dia` | quilos de ração fornecidos por vaca, por dia |

Os dados estão **embutidos na célula abaixo** em formato CSV (valores separados por vírgula) — assim não dependemos de nenhum link externo. Em projetos reais, você carregaria um arquivo com `pd.read_csv("caminho/do/arquivo.csv")`.

```python
# O texto abaixo é um CSV: a primeira linha traz os nomes das colunas,
# e cada linha seguinte é uma fazenda.
dados_csv = """fazenda,municipio,n_vacas,racao_kg_vaca_dia,temp_media_c,producao_litros_dia
F01,Patos de Minas,45,5.5,21.3,823
F02,Patos de Minas,120,7.2,22.1,2450
F03,Patos de Minas,60,4.0,23.5,880
F04,Patos de Minas,85,6.5,20.8,1720
F05,Pompéu,38,3.5,24.2,515
F06,Pompéu,95,6.8,23.0,1840
F07,Pompéu,52,5.0,22.6,895
F08,Pompéu,140,8.0,21.9,3050
F09,Lavras,70,6.0,20.5,1370
F10,Lavras,33,4.2,21.0,530
F11,Lavras,110,7.5,19.8,2460
F12,Lavras,58,5.8,20.2,1105
F13,Uberaba,66,5.2,24.8,1060
F14,Uberaba,150,8.5,23.9,3280
F15,Uberaba,42,3.8,25.1,575
F16,Uberaba,90,6.3,24.0,1685
F17,Sete Lagoas,75,6.6,22.4,1450
F18,Sete Lagoas,48,4.5,22.9,760
F19,Sete Lagoas,130,7.8,21.5,2870
F20,Sete Lagoas,55,5.4,21.1,1010
F21,Passos,62,5.9,22.0,1140
F22,Passos,105,7.0,21.7,2180
F23,Passos,36,4.8,23.2,600
F24,Passos,80,6.1,22.3,1490"""

# pd.read_csv transforma o texto em um DataFrame — a "planilha inteligente" do pandas.
df = pd.read_csv(io.StringIO(dados_csv))
print("Dados carregados!")
```

## 3. Estatísticas descritivas

O método `describe()` resume cada coluna numérica: média (`mean`), desvio padrão (`std`), mínimo, máximo e quartis.

**Pare e pense antes de executar:** quantas fazendas você acha que há por município? Qual deve ser a produção média? Depois compare com o resultado.

```python
# value_counts() conta quantas linhas existem para cada valor de uma coluna
print("Fazendas por município:")
print(df["municipio"].value_counts())

# describe() resume as colunas numéricas (round(1) arredonda para 1 casa decimal)
df.describe().round(1)
```

## 4. Visualizando os dados

Números resumem, mas **gráficos revelam**. Vamos criar dois:

1. **Gráfico de dispersão** (scatter): cada ponto é uma fazenda — ração por vaca no eixo X, produção no eixo Y. Se os pontos "sobem juntos", há relação entre as variáveis.
2. **Histograma**: mostra como a produção se distribui — muitas fazendas pequenas? Poucas gigantes?

```python
fig, eixos = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico 1: dispersão ração x produção
eixos[0].scatter(df["racao_kg_vaca_dia"], df["producao_litros_dia"], color="#6D28D9")  # roxo da Academy
eixos[0].set_xlabel("Ração (kg por vaca/dia)")
eixos[0].set_ylabel("Produção (litros/dia)")
eixos[0].set_title("Ração x Produção — cada ponto é uma fazenda")

# Gráfico 2: histograma da produção
eixos[1].hist(df["producao_litros_dia"], bins=8, color="#0D9488", edgecolor="white")  # teal da Academy
eixos[1].set_xlabel("Produção (litros/dia)")
eixos[1].set_ylabel("Número de fazendas")
eixos[1].set_title("Distribuição da produção diária")

plt.tight_layout()
plt.show()
```

## 5. Correlação: quem anda junto com quem?

A **correlação** mede, de -1 a +1, o quanto duas variáveis variam juntas:

- perto de **+1** → quando uma sobe, a outra sobe (correlação positiva forte);
- perto de **-1** → quando uma sobe, a outra desce (correlação negativa forte);
- perto de **0** → sem relação linear aparente.

**Atenção (isso cai no quiz!):** correlação **não** é causalidade — duas variáveis podem andar juntas por causa de um terceiro fator. Vamos explorar isso a fundo no Módulo 2.

Execute a célula e responda: **qual variável tem a correlação mais forte com `producao_litros_dia`?**

```python
# corr() calcula a correlação entre todas as colunas numéricas.
# numeric_only=True ignora colunas de texto (fazenda, municipio).
df.corr(numeric_only=True).round(2)
```
